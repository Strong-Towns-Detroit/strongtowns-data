# Explore Detroit BZA cases
This notebook reads prepared data. First use fetches it; subsequent runs reuse the local version. Set VERSION to the version printed below to reproduce an analysis.


```python
from strongtowns_data import bza
VERSION = None
dataset = bza.open(version=VERSION)
dataset.status()
```

```python
setbacks = dataset.cases(search="side setback")
print(f"{len(setbacks)} distinct cases")
setbacks[["case_history_id", "location", "proposal", "final_outcome"]]
```

## Cases and hearings
Continued hearings belong to one case. Count case histories for an application-level comparison. Missing classifications and unmatched locations remain unknown.


```python
print(f"{len(dataset.cases())} cases; {len(dataset.hearings())} hearing occurrences")
dataset.export("side-setbacks.csv", search="side setback")
```

```python
import matplotlib.pyplot as plt
counts = setbacks["final_outcome"].value_counts().sort_values()
fig, ax = plt.subplots(figsize=(8, 5))
counts = counts.rename(index={"granted_reversed": "Granted / reversed", "denied_upheld": "Denied / upheld", "dismissed_withdrawn": "Dismissed / withdrawn", "procedural_unresolved": "Unresolved", "mixed_or_other_decided": "Mixed / other"})
counts.plot.barh(ax=ax, color="#0072ce")
ax.set(title="BZA cases mentioning side setbacks", xlabel="Distinct cases", ylabel="Recorded outcome")
fig.tight_layout()
fig.savefig("side-setbacks.png", dpi=200)
```
